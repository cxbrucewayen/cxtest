```python
# 去掉空格,得到连续的十六进制字符串
hex_string = "E4BDA0E5A5BDEFBC81E4BB8AE5A4A9E6808EE4B988E6A0B7EFBC9FE69C89E4BB80E4B988E68891E58FAFE4BBA5E5B8AEE4BDA0E79A84E59097EFBC9F"
# 将十六进制字符串转换为二进制数据
binary_data = bytes.fromhex(hex_string)

# 将二进制数据按UTF-8解码
decoded_text = binary_data.decode('utf-8')

print(decoded_text)
```

```python
def analyze_hex(hex_string):
    # 移除空格
    hex_string = hex_string.replace(" ", "")
    

    # 将十六进制字符串转换为二进制数据
    binary_data = bytes.fromhex(hex_string)

    # 将二进制数据按UTF-8解码
    decoded_text = binary_data.decode('utf-8')

    print(decoded_text)
    


# 使用函数
hex_string = "E4 BD A0 E5 A5 BD EF BC 81 E5 BE 88 E9 AB 98 E5 85 B4 E4 B8 8E E6 82 A8 E4 BA A4 E6 B5 81 E3 80 82 E6 9C 89 E4 BB BB E4 BD 95 E9 97 AE E9 A2 98 E6 88 96 E9 9C 80 E8 A6 81 E5 B8 AE E5 8A A9 E7 9A 84 E5 90 B0 E6 96 B9 EF BC 8C E8 AF B7 E9 9A 8F E6 97 B6 E5 91 8A E8 AF 89 E6 88 91 E3 80 82"
analyze_hex(hex_string)
```
